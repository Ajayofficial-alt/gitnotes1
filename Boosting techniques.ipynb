```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,roc_curve,roc_auc_score
```

```python
credit_details=pd.read_csv(filepath_or_buffer="credit_card_clean.csv")
credit_details
```

#### DATA UNDERSTANDING

```python
credit_details.shape
```

```python
credit_details.isna().sum()
```

```python
credit_details.dtypes
```

```python
credit_details.describe()
```

####  DATA PREPARATION

##### DATA CLEANING

```python
del credit_details["ID"]
```

```python
credit_details
```

```python
le=LabelEncoder()
credit_details["SEX"]=le.fit_transform(credit_details["SEX"])
credit_details["EDUCATION"]=le.fit_transform(credit_details["EDUCATION"])
credit_details["MARRIAGE"]=le.fit_transform(credit_details["MARRIAGE"])
```

```python
credit_details.dtypes
```

##### DATA SEGMENTATION

```python
x=credit_details.drop(labels="DEFAULT",axis=1)
x
```

```python
y=credit_details["DEFAULT"]
y
```

#### MODEL BUILDING 

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=25,shuffle=True,stratify=y)
x_train,x_test,y_train,y_test
```

```python
x_train.shape,x_test.shape
```

```python
y_train.shape,y_test.shape
```

#### MODEL TRAINING

```python
adb_classifier=AdaBoostClassifier()
gb_classifier=GradientBoostingClassifier()
xgb_classifier=XGBClassifier()
lgm_classifier=LGBMClassifier()
```

```python
%%time
adb_classifier.fit(x_train,y_train)
```

```python
%%time
gb_classifier.fit(x_train,y_train)
```

```python
%%time
xgb_classifier.fit(x_train,y_train)
```

```python
%%time
lgm_classifier.fit(x_train,y_train)
```

#### MODEL TESTING

```python
y_pred_adb=adb_classifier.predict(x_test)
y_pred_gb=gb_classifier.predict(x_test)
y_pred_xgb=xgb_classifier.predict(x_test)
y_pred_lgb=lgm_classifier.predict(x_test)
```

#### MODEL EVALUATION

###### ADBOOSTING 

```python
accuracy_score(y_test,y_pred_adb)
```

```python
confusion_matrix(y_test,y_pred_adb)
```

```python
print(classification_report(y_test,y_pred_adb))
```

###### GRADIENT BOOSTING

```python
accuracy_score(y_test,y_pred_gb)
```

```python
confusion_matrix(y_test,y_pred_gb)
```

```python
print(classification_report(y_test,y_pred_gb))
```

###### EXTREMEGRADIENT BOOSTING 

```python
accuracy_score(y_test,y_pred_xgb)
```

```python
confusion_matrix(y_test,y_pred_xgb)
```

```python
print(classification_report(y_test,y_pred_xgb))
```

###### LIGHT GRADIENT BOOSTING

```python
round(accuracy_score(y_test,y_pred_lgb),ndigits=4)
```

```python
confusion_matrix(y_test,y_pred_lgb)
```

```python
print(classification_report(y_test,y_pred_lgb))
```

```python
adb_feature_imp=adb_classifier.feature_importances_
gb_feature_imp=gb_classifier.feature_importances_
xgb_feature_imp=xgb_classifier.feature_importances_
lgb_feature_imp=lgm_classifier.feature_importances_
```
