# Handling imbalance dataset

```python
import pandas as pd
```

```python
from sklearn import datasets
breast_cancer =datasets.load_breast_cancer()
breast_cancer
```

```python
breast_cancer_data=pd.DataFrame(breast_cancer.data,columns=breast_cancer.feature_names)
breast_cancer_data
```

```python
print(breast_cancer.DESCR)
```

#### DATA UNDERSTANDING

```python
breast_cancer_data.shape
```

```python
breast_cancer_data.isna().sum()
```

```python
breast_cancer_data.dtypes
```

```python
breast_cancer_data.describe()
```

#### DATA PREPARATION

```python
breast_cancer_data["target"]=breast_cancer.target
breast_cancer_data
```

```python
x=breast_cancer_data.drop(labels=["target"],axis=1)
x
```

```python
y=breast_cancer_data["target"]
y
```

#### MODEL BUILDING

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20, random_state=25,shuffle=True,stratify=y)
x_train,x_test,y_train,y_test
```

```python
rf_classifier=RandomForestClassifier(n_estimators=75, criterion='gini',max_depth=5)
rf_classifier
```

#### MODEL TRAINING

```python
rf_classifier.fit(x_train,y_train)
```

#### MODEL TESTING

```python
y_pred_train=rf_classifier.predict(x_train)
y_pred_train
```

```python
y_pred_test=rf_classifier.predict(x_test)
y_pred_test
```

#### MODEL EVALUATION

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
print("Training accuracy:",accuracy_score(y_train,y_pred_train))
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

###### TEST ACCURACY

```python
print("Test accuracy:",accuracy_score(y_test,y_pred_test))
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```
