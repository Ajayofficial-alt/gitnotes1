```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.feature_selection import RFE,SelectFromModel
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
breast_cancer=load_breast_cancer()
breast_cancer
```

```python
breast_cancer_data=pd.DataFrame(data=breast_cancer.data,columns=breast_cancer.feature_names)
breast_cancer_data
```

#### DATA PREPARATION

```python
breast_cancer_data["target"]=breast_cancer.target
breast_cancer_data
```

```python
x=breast_cancer_data.drop(labels=["target"],axis=1)
x
```

```python
y=breast_cancer_data["target"]
y
```

#### DATA SPLITTING 

```python
x_train,x_test,y_train,y_test=train_test_split(x,y, random_state=123,shuffle=True,stratify=y)
x_train,x_test,y_train,y_test
```

#### MODEL BUILDING

```python
rf_classifier=RandomForestClassifier(n_estimators=100,criterion='gini',max_depth=None)
```

#### MODEL TRAINING

```python
rf_classifier.fit(x_train,y_train)
```

#### MODEL TESTING

###### TRAINING ACCURACY

```python
y_pred_train=rf_classifier.predict(x_train)
y_pred_train
```

```python
y_pred_test=rf_classifier.predict(x_test)
y_pred_test
```

#### MODEL EVALUATION

###### Training accuracy

```python
accuracy_score(y_train,y_pred_train)
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```
