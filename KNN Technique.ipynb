## KNN Technique

#### BUSINESS UNDERSTANDING

```python
import pandas as pd
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.neighbors import KNeighborsClassifier
import warnings
warnings.filterwarnings("ignore")
from matplotlib import pyplot as plt
```

```python
fav_wine=datasets.load_wine()
wine_dataset=pd.DataFrame(data=fav_wine.data,columns=fav_wine.feature_names)
wine_dataset["target"]=fav_wine.target
wine_dataset
```

#### DATA UNDERSTANDING

```python
wine_dataset.dtypes
```

```python
wine_dataset.isna().sum()
```

```python
wine_dataset.describe()
```

#### DATA PREPARATION

```python
x=wine_dataset.drop(labels="target",axis=1)
y=wine_dataset["target"]
x
```

```python
y
```

#### MODEL BUILDING

```python
stand_scaler=StandardScaler()
stand_X=stand_scaler.fit_transform(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(stand_X,y,test_size=0.20,random_state=25,shuffle=True,stratify=y)
x_train,x_test,y_train,y_test
```

```python
x_train.shape,x_test.shape
```

```python
y_train.shape,y_test.shape
```

#### MODEL TRAINING

```python
knn_model=KNeighborsClassifier(n_neighbors=3)
knn_model.fit(x_train,y_train)
```

#### MODEL TESTING

```python
y_pred_train=knn_model.predict(x_train)
y_pred_train
```

```python
y_pred_test=knn_model.predict(x_test)
y_pred_test
```

#### MODEL EVALUATION

### K=3

```python
knn_model=KNeighborsClassifier(n_neighbors=3)
knn_model.fit(x_train,y_train)
y_pred_train=knn_model.predict(x_train)
print( 'Training_accuracy:', accuracy_score(y_train,y_pred_train))
```

### K=5

```python
knn_model=KNeighborsClassifier(n_neighbors=5)
knn_model.fit(x_train,y_train)
y_pred_train=knn_model.predict(x_train)
print( 'Training_accuracy:', accuracy_score(y_train,y_pred_train))
```

### K=9

```python
knn_model=KNeighborsClassifier(n_neighbors=9)
knn_model.fit(x_train,y_train)
y_pred_train=knn_model.predict(x_train)
print( 'Training_accuracy:', accuracy_score(y_train,y_pred_train))
```

### K=11

```python
knn_model=KNeighborsClassifier(n_neighbors=11)
knn_model.fit(x_train,y_train)
y_pred_train=knn_model.predict(x_train)
print( 'Training_accuracy:', accuracy_score(y_train,y_pred_train))
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
confusion_matrix(y_train,y_pred_train)
```

###### TEST ACCURACY

#### K=11

```python
knn_model=KNeighborsClassifier(n_neighbors=11)
knn_model.fit(x_train,y_train)
y_pred_test=knn_model.predict(x_test)
print( 'Testing_accuracy:', round(accuracy_score(y_test,y_pred_test),ndigits=4))

print(classification_report(y_train,y_pred_train))
print("confusion_matrix:",confusion_matrix(y_test,y_pred_test))
```

#### K=5

```python
knn_model=KNeighborsClassifier(n_neighbors=5)
knn_model.fit(x_train,y_train)
y_pred_test=knn_model.predict(x_test)
print( 'Testing_accuracy:', accuracy_score(y_test,y_pred_test))

print(classification_report(y_train,y_pred_train))
print("confusion_matrix:",confusion_matrix(y_test,y_pred_test))
```

#### K=9

```python
knn_model=KNeighborsClassifier(n_neighbors=9)
knn_model.fit(x_train,y_train)
y_pred_test=knn_model.predict(x_test)
print( 'Testing_accuracy:', accuracy_score(y_test,y_pred_test))

print(classification_report(y_train,y_pred_train))
print("confusion_matrix:",confusion_matrix(y_test,y_pred_test))
```

#### KFOLD

```python
from sklearn.model_selection import cross_val_score,KFold
```

```python
k_fold=KFold(n_splits=5,shuffle=True,random_state=23)
cross_score=cross_val_score(estimator=knn_model,X=stand_X,y=y,cv=k_fold)
print(cross_score)
print("Overall accuracy:",cross_score.mean().round(4))
print("Overall standard devation:",cross_score.std().round(4)) 
```
