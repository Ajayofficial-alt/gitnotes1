## Machine learning in linear regression

#### BUSINESS UNDERSTANDING

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
insurance_details=pd.read_csv(filepath_or_buffer="insurance data.csv")
insurance_details
```

#### DATA UNDERSTANDING 

```python
insurance_details.shape
```

```python
insurance_details.isna().sum()
```

```python
insurance_details.dtypes
```

```python
insurance_details.describe()
```

#### DATA PREPARATION

```python
insurance_details.dropna(inplace=True)
insurance_details
```

```python
x=insurance_details.drop(labels=["CASENUM","ATTORNEY"],axis=1)
x
```

```python
y=insurance_details[["ATTORNEY"]]
y
```

```python
x.shape
```

```python
x.isna().sum()
```

```python
y.shape
```

```python
y.isna().sum()
```

#### MODEL BUILDING

```python
from sklearn.linear_model import LogisticRegression
logistic_model=LogisticRegression()
```

#### MODEL TRAINING

```python
logistic_model.fit(x,y)
```

#### MODEL TESTING 

```python
logistic_model.intercept_
```

```python
logistic_model.coef_
```

```python
logistic_model.fit_intercept
```

```python
y_pred=logistic_model.predict(x)
print(y_pred)
```

#### MODEL EVALUATION

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_true=y, y_pred=y_pred)
```

#### MODEL DEPLOYMENT

```python
from pickle import dump
dump(obj=logistic_model,file=open("claimants.pkl","wb"))
from pickle import load
linear_intel_pkl = load(file=open("claimants.pkl","rb"))
print(linear_intel_pkl.predict(x))
```
