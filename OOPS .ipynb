# OOPS

```python
class official_doc:
    name         = None
    password_pin = None

    def __init__(self,name,password_pin):
        self.name = name
        self.password_pin = password_pin
        
    def check_pin(self,pin):
        if pin == self.password_pin:
            print("access granted")
        else:
            print("access denied")
            
    def login(self):
        print(f"{self.name} has logged in")

class top_level(official_doc):
    def main_doc(self):
           print(f"{self.name}higher official opened")

class mid_level(official_doc):
    def limit_access(self):
            print(f"{self.name}mid level people accessed")
    
```

```python
arjun = top_level ("arjun",1514)
priya = mid_level ("priya",1000)
```

```python
arjun.login()
arjun.main_doc()
arjun.check_pin(1514)
```

```python
priya.check_pin(1000)
priya.limit_access()
```

## Abastraction & inheritance

```python
from abc import ABC

class official_doc(ABC):
    
    def __init__(self,name,password_pin):
        self.name = name
        self.password_pin = password_pin

    def login(self):
        print(f"{self.name} has logged in")
    def access_doc(self):
        pass

class top_level(official_doc):
    def access_doc(self):
        print(f"{self.name} accessed all confidential documents")

class mid_level(official_doc):
    def access_doc(self):
        print(f"{self.name}accessed limited documents")
```

```python
arjun = top_level ("arjun",1514)
priya = mid_level ("priya",1000)
```

```python
arjun.login()
arjun.access_doc()
```

```python
priya.login()
priya.access_doc()
```

## polymorphism

```python
arjun = top_level("arjun",1514)
priya = mid_level("priya",1000)

employees  = [arjun,priya]

for emp in employees:
    emp.login()
    emp.access_doc()
```
