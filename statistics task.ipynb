##### 1.	The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance.

```python
from statsmodels.stats.weightstats import ztest
```

```python
sample = np.random.normal(68,10,50,)
z_stat,p_value =ztest(sample,value=65,alternative="larger")

print("Z statistic:", z_stat)
print("p_value:", p_value)

los= 0.05
if p_value < 0.05:
   print("reject Ho")
else:
   print("we donot reject Ho")

```

##### 5.Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods

```python
from scipy.stats import ttest_ind

method_A = np.random.normal(75, 12, 30)
method_B = np.random.normal(70, 10, 28)

t_stat, p_value = ttest_ind(method_A, method_B)

print("T Statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("reject Ho")
else:
    print("we donot reject Ho")
```

##### 9.A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs.



```python
from scipy.stats import f_oneway

program_A = np.random.normal(80, 8, 25)
program_B = np.random.normal(75, 7, 30)
program_C = np.random.normal(82, 6, 28)

f_stat, p_value = f_oneway(program_A, program_B, program_C)

print("F Statistic:", f_stat)
print("P-value:", p_value)

los = 0.05

if p_value < 0.05:
    print("reject Ho")
else:
    print("we donot reject Ho")
```

##### 10.A company wants to test whether gender and product preference are independent. The survey results are as follows:
●	Male: 60 prefer Product A, 40 prefer Product B
●	Female: 50 prefer Product A, 50 prefer Product B
At the 5% level of significance, test for independence.


```python
from scipy.stats import chi2_contingency
table = np.array([[60, 40],[50, 50]])

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:")
print(expected)

los = 0.05

if p_value < 0.05:
    print("Reject Ho")
    print("Gender and product preference are associated.")
else:
    print("we donot reject Ho")
    print("Gender and product preference are independent.")
```

##### 12.A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
●	Face 1: 18
●	Face 2: 22
●	Face 3: 20
●	Face 4: 25
●	Face 5: 17
●	Face 6: 18
At the 5% level of significance, test whether the die is fair.


```python
from scipy.stats import chisquare

observed = [18, 22, 20, 25, 17, 18]

expected = [20, 20, 20, 20, 20, 20]

chi2, p_value = chisquare(f_obs=observed, f_exp=expected)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)

los = 0.05

if p_value < 0.05:
    print("Reject Ho")
    print("The die is not fair.")
else:
    print("we donot reject Ho")
    print("There is no evidence that the die is unfair.")
```
